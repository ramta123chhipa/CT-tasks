```python
print("before cleaning")
df.isnull().sum()
```

```python
df['what_customers_said'].fillna('No review', inplace = True)
```

```python
df['seller_name'].fillna('Unknown Seller', inplace=True)
```

```python
df['videos'].fillna('[]', inplace=True)
```

```python
df['seller_information'].fillna('{}', inplace=True)
```

```python
df['variations'].fillna('[]', inplace=True)
```

```python
df['discount'] = df['discount'].fillna(0)
```

```python
print("after cleaning")
df.isnull().sum()
```

```python
print("selecting columns")
print(df[['title','rating','final_price','initial_price','discount','what_customers_said','seller_name','videos','seller_information','variations']].head())
```

```python
print("filtering data")
high_rating = df[df['rating']>4.0]

print(high_rating.head())
```

```python
print("Shape after removing duplicates:")
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df.shape)
```

```python
print("converting datatypes")
df['final_price'] = df['final_price'].astype(str)
df['final_price'] = df['final_price'].str.replace('₹', '', regex=False)
df['final_price'] = df['final_price'].str.replace(',', '', regex=False)
df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')
```

```python
print(" (deriving new column)calculating total amount")
df['total_amount'] = df['final_price'] * (df['discount'] / 100)
```

```python
print(df[['final_price', 'discount', 'total_amount']].head())
```
